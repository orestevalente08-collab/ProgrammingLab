# 4. Rispondere alle domande
Trasforma le seguenti richiesti in formule per estrarre infromazioni dal dataframe
- Mostrami le vendite maggiori di 10

- Mostrami i dati del 2018

- Mostrami le vendite maggiori di 13 e l'anno è il 2018

- Mostrami tutto TRANNE i casi in cui le vendite sono maggiori di 13 e l'anno è il 2018

- Mostrami i dati dove le vendite divise per 3 sono maggiori di 3

- Mostrami i dipendenti i cui nomi sono alfabeticamente dopo la J



```python
import pandas as pd

sales = pd.DataFrame(
    data={
        "employee": [
            "Katrina",
            "Guanyu",
            "Jan",
            "Roman",
            "Jacqueline",
            "Paola",
            "Esperanza",
            "Alaina",
            "Egweyn",
        ],
        "sales": [14, 17, 6, 12, 8, 3, 7, 15, 5],
        "year": [2018, 2019, 2020, 2018, 2020, 2019, 2019, 2020, 2020],
    }
)
sales
```

```python
sales.query('employee > "J"')   #ordine alfabetico dopo la J
sales.query('sales > 10')   #vendite maggiori di 10
sales.query('year == 2018') #vendite del 2018
sales.query('year == 2018 and sales >13')
sales.query('not(year == 2018 and sales >13)')
sales.query('sales/3 > 3')
```

# 5. Analisi di un dataset di video giochi
1. Quanti videogiochi sono stati pubblicati
2. Quali sono i generi più popolari? Mostralo con un bar plot
3. Plotta l'evoluzione del numero di giochi pubblicati nel tempo
4. Prova a riprodurre questo plot 




```python
# Import dei pacchetti
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Caricamento del dataset
url = 'https://zenodo.org/record/5898311/files/vgsales.csv'
df = pd.read_csv(url)
df.head()
```

```python
df.Name.count() #numero di giochi pubblicati


sns.barplot(
    data = df,
    x = "Genre",
    y = "Global_Sales"
)
plt.title("Popolarità dei generi")
plt.xlabel("Generi")
plt.ylabel("Vendite globali")
plt.show()
```

```python
sns.kdeplot(    #grafico di densità, violin plot
    data = df,
    x = "Year"
)
plt.title("Vendite di videogiochi nel tempo")
plt.show()
```

```python
Generi = df[df["Genre"].isin(["Action", "Misc", "Role-Playing", "Shooter", "Sports"])]
vendite_genere = Generi.groupby("Genre")[["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]].sum() #si crea prima la tabella

vendite_genere.plot(    #poi se ne fa il grafico
    kind="bar",
    stacked=True,
    colormap="Set1"
)
plt.title("Vendite per regione per genere")
plt.xlabel("Milioni di unità")
plt.ylabel("Genere")
plt.tight_layout()
plt.show()
```
